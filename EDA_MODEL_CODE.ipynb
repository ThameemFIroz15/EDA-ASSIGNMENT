```python

from google.colab import files
import pandas as pd
import io

uploaded = files.upload()
```

```python

filename = next(iter(uploaded))

if filename.lower().endswith(".csv"):
    df = pd.read_csv(io.BytesIO(uploaded[filename]))
elif filename.lower().endswith((".xlsx", ".xls")):
    df = pd.read_excel(io.BytesIO(uploaded[filename]))
else:
    raise ValueError("Please upload a CSV or Excel file.")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

display(df.head())
```

```python

print("Dataset information:")
df.info()

print("\nMissing values in each column:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nStatistical summary:")
display(df.describe(include="all").T)
```

```python

print("Available columns:")
for column in df.columns:
    print(f"{column}: {df[column].dtype}")

print("\nNumber of unique values in each column:")
print(df.nunique())
```

```python

print("Choose the column you want to predict:")
for i, column in enumerate(df.columns):
    print(i, ":", column)
```

```python

print("Revenue distribution:")
print(df["Revenue"].value_counts())

print("\nRevenue percentages:")
print((df["Revenue"].value_counts(normalize=True) * 100).round(2))

import matplotlib.pyplot as plt

df["Revenue"].value_counts().sort_index().plot(
    kind="bar", rot=0
)
plt.title("Distribution of Purchase Outcomes")
plt.xlabel("Revenue (0 = No Purchase, 1 = Purchase)")
plt.ylabel("Number of Sessions")
plt.show()
```

```python

X = df.drop(columns=["Revenue"])
y = df["Revenue"].astype(int)

print("Feature shape:", X.shape)
print("Target shape:", y.shape)
print("\nTarget classes:")
print(y.value_counts())
```

```python

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])

print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True).round(3))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True).round(3))
```

```python

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )
}

for name, model in models.items():
    model.fit(X_train, y_train)
    print(name, "trained successfully!")
```

```python

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)
import pandas as pd

results = []
predictions = {}

for name, model in models.items():
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    predictions[name] = y_pred

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(
            y_test, y_pred, zero_division=0
        ),
        "Recall": recall_score(
            y_test, y_pred, zero_division=0
        ),
        "F1-Score": f1_score(
            y_test, y_pred, zero_division=0
        ),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    })

results_df = pd.DataFrame(results)
print("Model Comparison:")
display(results_df.round(4))
```

```python

for name, model in models.items():
    print("\n" + "=" * 50)
    print(name)
    print("=" * 50)

    print("\nClassification Report:")
    print(classification_report(
        y_test,
        predictions[name],
        target_names=["No Purchase", "Purchase"],
        zero_division=0
    ))

    print("Confusion Matrix:")
    print(confusion_matrix(y_test, predictions[name]))
```

```python

results_df.set_index("Model")[
    ["Accuracy", "Precision", "Recall", "F1-Score", "ROC-AUC"]
].plot(kind="bar", figsize=(12, 6))

plt.title("Comparison of Machine Learning Models")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=15)
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()
```

```python

best_model_name = results_df.sort_values(
    by="F1-Score", ascending=False
).iloc[0]["Model"]

best_model = models[best_model_name]

print("Best model based on test F1-score:", best_model_name)

display(
    results_df[results_df["Model"] == best_model_name].round(4)
)
```

```python

rf_model = models["Random Forest"]

importance = pd.Series(
    rf_model.feature_importances_,
    index=X_train.columns
).sort_values(ascending=False)

print("Top 10 important features:")
display(importance.head(10).to_frame("Importance"))

importance.head(10).sort_values().plot(
    kind="barh", figsize=(9, 6)
)
plt.title("Top 10 Feature Importances - Random Forest")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()
```

```python

import joblib

joblib.dump(best_model, "best_revenue_model.pkl")

print("Model saved successfully!")
```
